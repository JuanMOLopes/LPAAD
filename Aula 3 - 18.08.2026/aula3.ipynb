Exercício 1:

```python
tempos = [32, 48, 27, 55, 41, 29]

tempo_maximo = 40

tempo_total = 0
entregas_atrasadas = 0
quantidade_entregas = len(tempos)

for i in range(quantidade_entregas):
    status = ""
    if tempos[i] > tempo_maximo:
        status = "com atraso"
        entregas_atrasadas += 1
    else:
        status = "no prazo"
    
    tempo_total += tempos[i]
    
    print(f"{'=' * 55}")
    print(f"Entrega {i + 1}: Entregue em {tempos[i]} minutos | Entregue {status}\n")
    
print(f"{'=' * 55}\n")
print(f"RELATÓRIO")
print(f"Tempo total das entregas: {tempo_total} minutos")
print(f"Tempo médio das entregas: {tempo_total / quantidade_entregas:.0f} minutos")
print(f"Entregas atrasadas: {entregas_atrasadas} ({(entregas_atrasadas/quantidade_entregas)*100:.0f}%)")
    
```

```python
producoes = []

valor = 0

while valor != -1:
    if valor < 300 and valor > 800:
        print("Valor inválido. A quantidade produzida deve ser entre 300 e 800.")
    elif valor ==0:
        print("Iniciando o sistema...")
    else:
        producoes.append(valor)
        
    valor = int(input("Digite a quantidade produzida dentro do seu turno (300 a 800) ou -1 para encerrar: "))

meta = ""
producao_total = 0
atingiu_meta = 0
for producao in range(len(producoes)):
    producao_total += producoes[producao]
    
    if producoes[producao] >= 500:
        atingiu_meta += 1
        print(f"Turno {producao + 1}: {producoes[producao]} unidades | Atingiu a meta")
    elif producoes[producao] >= 400:
        print(f"Turno {producao + 1}: {producoes[producao]} unidades | Abaixo da meta")
    else:
        print(f"Turno {producao + 1}: {producoes[producao]} unidades | Crítico")
        
print(f"{'=' * 55}\n")
print(f"RELATÓRIO\n")
print(f"Produção total: {producao_total:.0f} unidades")
print(f"Quantidade de turnos: {len(producoes)}")
print(f"Produção média: {producao_total / len(producoes):.0f} unidades")
print(f"Turnos que atingiram a meta: {atingiu_meta} ({(atingiu_meta/len(producoes))*100:.0f}%)")
```
