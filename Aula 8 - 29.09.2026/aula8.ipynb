```python
# Pratica - Cadastro de funcionário

def cadastrar_funcionarios(nome, idade, salario, setor):
    if not nome or type(nome) != str:
        print("Nome é obrigatório")
        return
      
    if idade < 18 or idade > 70:
        print("Idade inválida")
        return
    
    if salario < 0:
        print("Salário inválido")
        return
    
    setores = ["ti", "rh", "financeiro", "comercial"]
    if setor.lower() not in setores:
        print("Setor inválido")
        return
    
    print(f"Funcionário cadastrado com sucesso!")
    
cadastrar_funcionarios("Juan", 18, 5000.00, "TI")
cadastrar_funcionarios("", 18, 5000.00, "TI")
cadastrar_funcionarios("Juan", 80, 5000.00, "TI")
cadastrar_funcionarios("Juan", 18, -100, "TI")
cadastrar_funcionarios("Juan", 18, 5000.00, "Marketing")
```
