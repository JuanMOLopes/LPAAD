```python
def texto_colorido(texto, cor="normal", estilo='normal'):    
    match cor.lower():
        case "preto":
            codigo_cor = 30
        case "vermelho":
            codigo_cor = 31
        case "verde":
            codigo_cor = 32
        case "amarelo":
            codigo_cor = 33
        case "azul":
            codigo_cor = 34
        case "magenta":
            codigo_cor = 35
        case "ciano":
            codigo_cor = 36
        case "cinza claro":
            codigo_cor = 37
        case "cinza escuro":
            codigo_cor = 90
        case "vermelho claro":
            codigo_cor = 91
        case "verde claro":
            codigo_cor = 92
        case "amarelo claro":
            codigo_cor = 93
        case "azul claro":
            codigo_cor = 94
        case "magenta claro":
            codigo_cor = 95
        case "ciano claro":
            codigo_cor = 96
        case "branco claro":
            codigo_cor = 97
        case _:
            codigo_cor = 0  
    
    match estilo.lower():
        case 'negrito': 
            codigo_estilo = 1
        case 'italico':
            codigo_estilo = 3
        case 'sublinhado':
            codigo_estilo = 4
        case _:
            codigo_estilo = 0
            
    return f"\033[{codigo_estilo};{codigo_cor}m{texto}\033[0m"
            

print(f"Olá, {texto_colorido("João", "branco claro", "sublinhado")}, tudo bem?")
```

```python
def calcular_mediana(lista):
    lista = sorted(lista)
    print(f"Lista ordenada: {lista}")
    
    num_elementos = len(lista)
    meio = num_elementos // 2

    if num_elementos % 2 == 0:
       print(f"A mediana é: {(lista[meio-1] + lista[meio]) / 2}")
    else:
       print(f"A mediana é: {lista[meio]}") 
     
calcular_mediana([23,14,31,17,20])
calcular_mediana([75,1,4,3,19,50,64,31])

```

```python
def calcular_moda(lista):
    moda = [0]
    maior_frequencia = 0
    
    for valor in lista:
        frequencia = lista.count(valor)
        
        if frequencia > maior_frequencia:
            maior_frequencia  = frequencia
            moda = valor
            
        return moda
        
          
```

```python
def calcular_moda(valores):
    modas = []
    maior_frequencia = 0
    
    for valor in valores:
        frequencia = valores.count(valor)
        
        if frequencia > maior_frequencia:
            maior_frequencia  = frequencia
            modas = [valor]
        
        elif frequencia == maior_frequencia and valor not in modas:
            modas.append(valor)
            
        
    if maior_frequencia == 1:
        return "Amodal"
    
    elif len(modas) == 1:
        return "Unimodal", modas
    
    elif len(modas) == 2:
        return "Bimodal", modas
    
    else:
        return "Multimodal", modas        
          
```

```python
dados_1 = [10,20,20,30,40]
dados_2 = [10,10,20,20,30]
dados_3 = [10,10,20,20,30,30,40]
dados_4 = [10,20,30,40]

print(calcular_moda(dados_1))
print(calcular_moda(dados_2))
print(calcular_moda(dados_3))
print(calcular_moda(dados_4))
```
