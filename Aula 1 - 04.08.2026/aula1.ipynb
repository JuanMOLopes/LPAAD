## Problema 1: Desenvolvendo uma Calculadora Básica

```python
# Exiba uma mensagem explicando o funcionamento da calculadora
print(f"Calculadora Básica \n\n{"-" * 50}\n\nAo digitar dois números, você receberá os resultados das seguintes operações:\n\n\tSoma \n\tSubtração \n\tMultiplicação \n\tDivisão \n\tDivisão Inteira \n\tResto da Divisão \n\tPotência")

#Solicite ao usuário dois valores numéricos
num1 = float(input("Digite um número [1/2]"))
num2 = float(input("Digite um número [2/2]"))

# Calcule: Soma, Subtração, Multiplicação, Divisão, Divisão Inteira, Resto da Divisão, Potência
soma = num1 + num2
subtracao = num1 - num2
multiplicacao = num1 * num2
divisao = num1 / num2
divisao_inteira = num1 // num2
resto_divisao = num1 % num2
potencia = num1 ** num2

# Exiba todos os resultados na tela
print(f"\n{"-" * 50}\n")
print(f"Resultados\n")
print(f"Soma: {soma}")
print(f"Subtração: {subtracao}")
print(f"Multiplicação: {multiplicacao}")
print(f"Divisão: {divisao}")
print(f"Divisão Inteira: {divisao_inteira}")
print(f"Resto da Divisão: {resto_divisao}")
print(f"Potência: {potencia}")
```

## Problema 2: Controle de Produção

```python
# Armazene a produção de cada dia em uma variável
segunda = 1240
terca = 1315
quarta = 1288
quinta = 1360
sexta = 1297

# Calcule a producao total da semana
total = segunda + terca + quarta + quinta + sexta

# Calcule a produção média diária
media = total / 5

# Determine quantas peças faltariam para atingir a meta semanal de 7000 peças
meta = 7000
falta_para_meta = meta - total

# Exiba todas as informações na tela
print(f"Relatório Semanal\n\n{"-" * 30}\n")
print(f'Total produzido: {total} peças')
print(f'Média diária produzida: {media} peças')
print(f'Quanto falta para a meta: {falta_para_meta} peças')
```

## Problema 4: Monitoramento de Temperatura Industrial

|Horário|Temperatura (°C)|
|-------|----------------|
|08:00|72,3|
|09:00|73,1|
|10:00|74,0|
|11:00|72,8|
|12:00|71,5|
|13:00|75,2|
|14:00|73,6|
|15:00|74,4|

```python
soma = 72.3 + 73.1 + 74.0 + 72.8 + 71.5 + 75.2 + 73.6 + 74.4

# Desenvolva um programa em Python que calcule:

# A temperatura média da máquina
media = soma / 8

# Exiba manualmente a temperatura máxima e mínima registrada
maxima = 75.2
minima = 71.5

# Calcule a amplitude térmica
amplitude = maxima - minima

# Considerando que a temperatura ideal de operação é 73 ºC,
# Calcule a diferença entre a temperatura média e a temperatura ideal;
diferenca = media - 73

# Exiba uma mensagem utilizando uma f-string, informando a temperatura média da máquina, a amplitude térmica e a diferença em relação à temperatura ideal, apresentando os valores com uma casa decimal
print(f"A temperatura média da máquina é {media:.1f} ºC, a amplitude térmica é {amplitude:.1f} ºC e a diferença em relação à temperatura ideal é {diferenca:.1f} ºC.")
```
