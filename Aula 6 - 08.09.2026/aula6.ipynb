## Cálculo da variância

```python
# Cálculo da média
def calcula_media(lista):
    soma = 0
    for i in lista:
        soma += i
    media = soma / len(lista)
    return media

# --------------------------------------------------------------
# Cálculo da variância
def calcula_variancia(lista):
    media = calcula_media(lista)
    
    soma_diferencas = 0
    for i in lista:
        diferenca = i - media
        diferenca_ao_quadrado = diferenca ** 2
        soma_diferencas += diferenca_ao_quadrado
        
    variancia = soma_diferencas / len(lista)
    return variancia

tempos = [12,18,15,21,14,16]
print(f"Variância: {calcula_variancia(tempos)}")
```

## Cálculo do desvio padrão

```python
# Cálculo do desvio padrão
def calcula_desvio_padrao(lista):
    variancia = calcula_variancia(lista)
    
    import math
    desvio_padrao = math.sqrt(variancia)
    return desvio_padrao

print(f"Desvio Padrão: {calcula_desvio_padrao(tempos)}")
```

## Cálculo da amplitude

```python
# Cálculo da amplitude
def calcula_amplitude(lista):
    maximo = max(lista)
    minimo = min(lista)
    amplitude = maximo - minimo
    return amplitude

valores = [18500, 22100, 19800, 24300, 25600, 21700]
print(f"Amplitude: {calcula_amplitude(valores)}")
```

## Cálculo de quartis

```python
def calcular_quartis(lista):
    lista_ordenada = sorted(lista)
    len_lista = len(lista)
    
    print(lista_ordenada)
    
    quartis = []
    if len_lista % 2 == 0:
        posicao_quartis = []
        
        for quartil in range(1, 4):
            posicao = (quartil * (len_lista + 1)) / 4
            inteiro_posicao = int(posicao)
            decimal_posicao = posicao - inteiro_posicao
            
            diferenca = lista_ordenada[inteiro_posicao] - lista_ordenada[inteiro_posicao - 1]
            
            proporcao = diferenca * decimal_posicao
            
            quartis.append(lista_ordenada[inteiro_posicao - 1] + proporcao)
            
    else:
        for quartil in range(1, 4):
            posicao = int((quartil * (len_lista + 1)) / 4)
            print(f"{quartil}º Quartil: {lista_ordenada[posicao-1]}")
            
    return quartis
            

vendas = [1,2,3,4,5,6,7]
quartis = calcular_quartis(vendas)

for i in range(1,4):
    print(f"{i}º Quartil: {quartis[i - 1]}")
```

## Cálculo do Coeficiente de Variação (CV)

```python
def calcula_coeficiente_variacao(lista):
    desvio_padrao = calcula_desvio_padrao(lista)
    media = calcula_media(lista)
    
    coeficiente_variacao = (desvio_padrao / media) * 100
    return coeficiente_variacao

valores = [98,101,100,99,102,100,101]
print(f"Coeficiente de variação: {calcula_coeficiente_variacao(valores):.2f}%")
```
