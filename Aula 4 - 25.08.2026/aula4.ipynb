Problema 1

```python
estoque = [18, 5, 0, 31, 4, 7, 0, 14, 3]

esgotados = 0
criticos = 0

total_produtos = len(estoque)

print(f"{' '*10}{'-'*10} ESTOQUE {'-'*10}")
print(f"{'-'*50}")

for i in range(total_produtos):
    if estoque[i] == 0:
        esgotados += 1
        print(f"Produto {i} | {estoque[i]} unidades em estoque - Esgotado")
    elif estoque[i] <= 5:
        criticos += 1
        print(f"Produto {i} | {estoque[i]} unidades em estoque - Crítico")
    else:
        print(f"Produto {i} | {estoque[i]} unidades em estoque - Disponível") 

print(f"{'-'*50}")

print(f"Quantidade de produtos esgotados: {esgotados}")
print(f"Quantidade de produtos com estoque crítico: {criticos}")
print(f"Quantidade de produtos com estoque crítico: {criticos}")
print(f"Percentual dos produtos que precisam de reposição: {(esgotados + criticos) / total_produtos * 100:.2f}")
```

Problema 3

```python
rede = [
    [120, 135, 128, 141], 
    [98, 105, 110, 103], 
    [150, 145, 160, 155], 
    [116, 161, 108, 121]
]
atendimentos = []

print(f"{'-'*75}")

teste = max(rede)
for unidade in range(len(rede)):
    total = sum(rede[unidade])
    atendimentos.append(total)
    print(f'A unidade {unidade + 1} possui um total de {total} atendimentos')

print(f"{'-'*75}")

maior_valor = max(atendimentos)
for unidade in range(len(atendimentos)):
    if atendimentos[unidade] == maior_valor:
        print(f'A unidade {unidade + 1} realizou o maior número de atendimentos')

abaixo_do_esperado = 0
lista_abaixo_do_esperado = []
maior = rede[0][0] 
for unidade in rede:
    for semana in unidade:
        if semana < 110:
            abaixo_do_esperado += 1
            lista_abaixo_do_esperado.append(semana)
        if semana > maior:
            maior = semana
            
print(f"{abaixo_do_esperado} registros semanais ficaram abaixo de 110 atendimentos: {lista_abaixo_do_esperado}")
print(f"A maior quantidade de atendimentos registrada em uma única semana foi de {maior}")
    

```

```python
produtos = {
    "Produto":"Notebook", 
    "Valor":4200.00, 
    "Estoque":8
}

print(f"{' '*10}{'-'*10} Keys {'-'*10}")
for i in produtos.keys():
    print(i)
    
print(f"\n{' '*10}{'-'*10} Values {'-'*10}")
for i in produtos.values():
    print(i)
    
print(f"\n{' '*10}{'-'*10} Items {'-'*10}")
for i in produtos.items():
    print(i)
```

```python
loja = [
    {"Produto": "Notebook", "Preco": 4200.00, "Estoque": 8},
    {"Produto": "Mouse", "Preco": 120.00, "Estoque": 25},
    {"Produto": "Monitor", "Preco": 1350.00, "Estoque": 4},
    {"Produto": "Teclado", "Preco": 280.00, "Estoque": 15},
    {"Produto": "Headset", "Preco": 450.00, "Estoque": 3},
]

print(f"\n{' '*10}{'-'*10} Produtos {'-'*10}")
for produto in loja:
    print(f"{produto["Produto"]} | R$ {produto["Preco"]:.2F} | {produto["Estoque"]} unidades disponíveis")

estoque_baixo = 0
print(f"\n{' '*10}{'-'*10} Estoque baixo {'-'*10}")
for produto in loja:
    if produto["Estoque"] < 10:
        estoque_baixo += 1
        print(f"{produto["Produto"]}: {produto["Estoque"]} unidades")
        
percentagem = (estoque_baixo / len(loja)) * 100
print(f"\nExistem {estoque_baixo} produtos com baixo estoque ({percentagem})%")
```
