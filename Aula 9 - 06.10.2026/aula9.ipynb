```python
def texto_colorido(texto, cor="normal", estilo='normal'):    
    match cor.lower():
        case "preto":
            codigo_cor = 30
        case "vermelho":
            codigo_cor = 31
        case "verde":
            codigo_cor = 32
        case "amarelo":
            codigo_cor = 33
        case "azul":
            codigo_cor = 34
        case "magenta":
            codigo_cor = 35
        case "ciano":
            codigo_cor = 36
        case "cinza claro":
            codigo_cor = 37
        case "cinza escuro":
            codigo_cor = 90
        case "vermelho claro":
            codigo_cor = 91
        case "verde claro":
            codigo_cor = 92
        case "amarelo claro":
            codigo_cor = 93
        case "azul claro":
            codigo_cor = 94
        case "magenta claro":
            codigo_cor = 95
        case "ciano claro":
            codigo_cor = 96
        case "branco claro":
            codigo_cor = 97
        case _:
            codigo_cor = 0  
    
    match estilo.lower():
        case 'negrito': 
            codigo_estilo = 1
        case 'italico':
            codigo_estilo = 3
        case 'sublinhado':
            codigo_estilo = 4
        case _:
            codigo_estilo = 0
            
    return f"\033[{codigo_estilo};{codigo_cor}m{texto}\033[0m"
```

## Prática 1: Operação Bancária

```python
def calculo_juros_simples(capital, taxa, tempo):
    juros = capital * (taxa / 100) * tempo
    montante = juros + capital
    return juros, montante

def calculo_juros_composto(capital, taxa, tempo):
    montante = capital * (pow((1 + (taxa/100)), tempo))
    juros = montante - capital
    return juros, montante 

try:
    valor_saque = float(input(f"Valor do saque:"))
    parcelas = int(input(f"Quantidade de parcelas:"))    
    tipo_juros = int(input(f"Tipo de juros [1 - Simples | 2 - Composto]:"))
    taxa_juros = float(input(f"Digite a taxa de juros mensal (%):"))
        
    
    limite_disponivel = 8000 - valor_saque
    valor_parcela = valor_saque / parcelas
    
    if valor_saque < 100 or valor_saque > 5000:
        print(f"{texto_colorido('[ERRO] Valor para saque inválido', 'vermelho', 'italico')}")
    
    elif limite_disponivel < 3500:
        print(f"{texto_colorido('[ERRO] Limite restante menor que R$3.500,00', 'vermelho', 'italico')}")
    
    elif parcelas < 1 or parcelas > 6:
        print(f"{texto_colorido('[ERRO] Número de parcelas inválido', 'vermelho', 'italico')}")
        
    elif valor_parcela < 50:
        print(f"{texto_colorido('[ERRO] Valor da parcela menor que R$50,00', 'vermelho', 'italico')}")
        
    elif tipo_juros not in [1, 2]:
        print(f"{texto_colorido('[ERRO] Tipo de juros inválido', 'vermelho', 'italico')}")
    
    elif taxa_juros < 0:
        print(f"{texto_colorido('[ERRO] Taxa de juros inválida', 'vermelho', 'italico')}")
    
    else:
        print(f"{texto_colorido('Operação realizada com sucesso!', 'verde', 'negrito')}")
        print(f"Valor do saque: {texto_colorido(f'R${valor_saque:.2f}', 'verde', 'negrito')}")
        print(f"Limite restante disponível: {texto_colorido(f'R${limite_disponivel:.2f}', 'verde', 'negrito')}")
        print(f"Quantidade de parcelas: {texto_colorido(f'{parcelas}', 'verde', 'negrito')}")
        print(f"Valor da parcela para pagamento: {texto_colorido(f'R${valor_parcela:.2f}', 'verde', 'negrito')}")
        if tipo_juros == 1:
            juros, montante = calculo_juros_simples(valor_saque, taxa_juros, parcelas)
            print(f"Juros simples: {texto_colorido(f'R${juros:.2f}', 'verde', 'negrito')}")
            print(f"Montante total a ser pago: {texto_colorido(f'R${montante:.2f}', 'verde', 'negrito')}")
        else:
            juros, montante = calculo_juros_composto(valor_saque, taxa_juros, parcelas)
            print(f"Juros compostos: {texto_colorido(f'R${juros:.2f}', 'verde', 'negrito')}")
            print(f"Montante total a ser pago: {texto_colorido(f'R${montante:.2f}', 'verde', 'negrito')}")
    
except ValueError:
    print(f'{texto_colorido("Valor digitado inválido!", "vermelho", "italico")}')
    
except ZeroDivisionError:
    print(f'{texto_colorido("Número de parcelas não pode ser zero!", "vermelho", "italico")}')
    
finally:
    print(f"{texto_colorido('Programa finalizado', 'azul', 'negrito')}")
```

## Prática 2: Validando *args

```python
def registrar_notas(aluno, *notas):
    if not aluno or type(aluno) != str:
        print(f"{texto_colorido(f'[ERRO] O nome do aluno é obrigatório', 'verde', 'negrito')}")
        return
    if len(notas) == 0:
        print(f"{texto_colorido(f'[ERRO] Nenhuma nota foi informada para o aluno {aluno}', 'verde', 'negrito')}")
        return
    else:
        media = sum(notas) / len(notas)
        print(f"Aluno: {texto_colorido(aluno, 'azul', 'negrito')}")
        print(f"Notas: {texto_colorido(notas, 'verde', 'negrito')}")
        print(f"Média: {texto_colorido(f'{media:.2f}', 'amarelo', 'negrito')}")
        
registrar_notas("João", 7.5, 8.0, 9.0)
        
```
