## Operadores Relacionais

|Operador|Significado|Exemplo|
|--------|-----------|-------|
|`>`|Maior que|5 > 2|
|`<`|Menor que|3 < 4|
|`>=`|Maior ou igual|5 >= 5|
|`<=`|Menor ou igual|3 <= 3|
|`==`|Igual a|7 == 7|
|`!=`|Diferente de|10 != 3|

## O que é o comando if?

```python
temperatura = float(input("Digite a temperatura atual: "))

if temperatura > 30:
    print("Está muito calor, hidrate-se")
```

## If e Else

```python
senha = 1234
senha_digitada = int(input("Digite a senha: "))

if senha_digitada == senha:
    print("Acesso permitido")
else:
    print("Acesso negado")
```

## Valores Booleanos

```python
# Solicite ao usuário o saldo disponível
saldo_disponivel = float(input("Digite o saldo disponível em conta: "))

# Solicite o valor da transação
valor_transacao = float(input("Digite o valor que deseja transferir: "))

# Crie uma variável booleana, saldo_suficiente, que indique se o saldo é suficiente para realizar a operação
saldo_suficiente = saldo_disponivel >= valor_transacao

# Exiba o resultado da variável booleana (True or False)
print(f"Saldo suficiente: {saldo_suficiente}")

# Utilize estrutura condicional (if/else) para exibir a mensagem de transação autorizada ou transação negada: saldo insuficienmte
if saldo_suficiente:
    print(f"Transação autorizada!")
else:
    print(f"Transação negada: saldo insuficiente")
```

## Operadores Lógicos
