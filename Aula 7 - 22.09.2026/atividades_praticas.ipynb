#### 1. Percorra a lista de chamados utilizando enumerate(), numerando os registros a partir de 1;

````python
for numero, chamado in enumerate(chamados, start=1):
    print(f"{numero} | {chamado}")
````

#### 2. Converta as datas de abertura e fechamento, atualmente armazenadas como strings, para objetos datetime;

````python
for chamado in chamados:
    chamado["abertura"] = datetime.strptime(chamado["abertura"], "%d/%m/%Y")
    chamado["fechamento"] = datetime.strptime(chamado["fechamento"], "%d/%m/%Y")
    
    print(chamado)
````

#### 3. Calcule o tempo de atendimento, em dias, de cada chamado;


````python
for numero, chamado in enumerate(chamados, start=1):
    diferenca = chamado["fechamento"] - chamado["abertura"]
    
    chamado["tempo_atendimento"] = diferenca.days
    
    print(f"Tempo de atendimento do chamado {numero}: {chamado['tempo_atendimento']} dias")
````

#### 4. Apresente, para cada chamado:
- número do registro;
- código;
- responsável;
- prioridade;
- tempo de atendimento em dias;
- situação do prazo.


#### 5. Classifique cada chamado como:
- Dentro do prazo: tempo de atendimento menor ou igual a 3 dias;
- Fora do prazo: tempo de atendimento maior que 3 dias.
  
Uma possível apresentação para cada registro é:

1 - CH001 | Maria | Alta | 2 dias | Dentro do prazo

````python
for numero, chamado in enumerate(chamados, start=1):
    if chamado["tempo_atendimento"] <= 3:
        chamado["situacao"] = "Dentro do prazo"
    else:
        chamado["situacao"] = "Fora do prazo"
        
    print(f"{numero} - {chamado["codigo"]} | {chamado["responsavel"]} | {chamado["prioridade"]} | {chamado["tempo_atendimento"]} dias | {chamado["situacao"]}")
````

### Parte 2: Resumo dos atendimentos

Após processar todos os chamados, apresente:

1. Quantidade total de chamados;
2. Quantidade de chamados dentro do prazo;
3. Quantidade de chamados fora do prazo;
4. Código do chamado que apresentou o maior tempo de atendimento;
5. Maior tempo de atendimento encontrado;
6. Quantidade de chamados de prioridade:
- Alta;
- Média;
- Baixa

````python
def resumo_atendimentos(chamados):
    print("RESUMO DOS ATENDIMENTOS\n")
    
    dentro_prazo = 0
    fora_prazo = 0
    maior_tempo = {"codigo": None, "tempo": 0}
    prioridade_alta = 0
    prioridade_media = 0
    prioridade_baixa = 0
    
    for chamado in chamados:
        if chamado["situacao"] == "Dentro do prazo":
            dentro_prazo += 1
        else:
            fora_prazo += 1
            
        if chamado["tempo_atendimento"] > maior_tempo["tempo"]:
            maior_tempo["codigo"] = chamado["codigo"]
            maior_tempo["tempo"] = chamado["tempo_atendimento"]
            
        if chamado["prioridade"] == "Alta":
            prioridade_alta += 1
        elif chamado["prioridade"] == "Média":
            prioridade_media += 1
        else:
            prioridade_baixa += 1

    print(f"Quantidade total de chamados: {len(chamados)}")
    print(f"Quantidade de chamados dentro do prazo: {dentro_prazo}")
    print(f"Quantidade de chamados fora do prazo: {fora_prazo}")
    print(f"Código do chamado com maior tempo de atendimento: {maior_tempo['codigo']}")
    print(f"Maior tempo de atendimento encontrado: {maior_tempo['tempo']} dias")
    print(f"Quantidade de chamados de prioridade Alta: {prioridade_alta}")
    print(f"Quantidade de chamados de prioridade Média: {prioridade_media}")
    print(f"Quantidade de chamados de prioridade Baixa: {prioridade_baixa}")
    
resumo_atendimentos(chamados)
````

### Parte 3: Organização por prioridade


````python
chamados_por_prioridade = {
    "Alta": [],
    "Média": [],
    "Baixa": []
}

for chamado in chamados:
    if chamado["prioridade"] == "Alta":
        chamados_por_prioridade["Alta"].append(chamado["codigo"])
    elif chamado["prioridade"] == "Média":
        chamados_por_prioridade["Média"].append(chamado["codigo"])
    elif chamado["prioridade"] == "Baixa":
        chamados_por_prioridade["Baixa"].append(chamado["codigo"])
        
print(chamados_por_prioridade)
````
