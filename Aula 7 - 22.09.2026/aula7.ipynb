## Bibliotecas e Funções

```python
import statistics
```

```python
def calcular_moda(valores):
    modas = []
    maior_frequencia = 0
    
    for valor in valores:
        frequencia = valores.count(valor)
        
        if frequencia > maior_frequencia:
            maior_frequencia  = frequencia
            modas = [valor]
        
        elif frequencia == maior_frequencia and valor not in modas:
            modas.append(valor)
            
        
    if maior_frequencia == 1:
        return "Amodal"
    
    elif len(modas) == 1:
        return "Unimodal", modas
    
    elif len(modas) == 2:
        return "Bimodal", modas
    
    else:
        return "Multimodal", modas        
```

```python
def calcula_amplitude(lista):
    maximo = max(lista)
    minimo = min(lista)
    amplitude = maximo - minimo
    return amplitude
```

```python
def calcula_coeficiente_variacao(lista):
    desvio_padrao = statistics.stdev(lista)
    media = statistics.mean(lista)
    
    coeficiente_variacao = (desvio_padrao / media) * 100
    return coeficiente_variacao
```

```python
def calcular_quartis(lista):
    lista_ordenada = sorted(lista)
    len_lista = len(lista)
    
    quartis = []
    if len_lista % 2 == 0:
        posicao_quartis = []
        
        for quartil in range(1, 4):
            posicao = (quartil * (len_lista + 1)) / 4
            inteiro_posicao = int(posicao)
            decimal_posicao = posicao - inteiro_posicao
            
            diferenca = lista_ordenada[inteiro_posicao] - lista_ordenada[inteiro_posicao - 1]
            
            proporcao = diferenca * decimal_posicao
            
            quartis.append(lista_ordenada[inteiro_posicao - 1] + proporcao)
            
    else:
        for quartil in range(1, 4):
            posicao = int((quartil * (len_lista + 1)) / 4)
            quartis.append(lista_ordenada[posicao-1])
            
    return quartis
```

## Exercício 1: Qual equipe teve melhor desempenho?

```python
equipe_a = [42, 38, 45, 41, 39, 44, 38]
equipe_b = [30, 52, 35, 48, 29, 55, 38]

print(f"{'=' * 30} Dados das Equipes  {'=' * 30}")

print(f"Média da equipe A: {statistics.mean(equipe_a)}")
print(f"Média da equipe B: {statistics.mean(equipe_b)}")

print(f"{'=' * 80}")

print(f"Mediana da equipe A: {statistics.median(equipe_a)}")
print(f"Mediana da equipe B: {statistics.median(equipe_b)}")

print(f"{'=' * 80}")

print(f"Moda da equipe A: {calcular_moda(equipe_a)}")
print(f"Moda da equipe B: {calcular_moda(equipe_b)}")

print(f"{'=' * 80}")

print(f"Amplitude da equipe A: {calcula_amplitude(equipe_a)}")
print(f"Amplitude da equipe B: {calcula_amplitude(equipe_b)}")
```

### Conclusões:

As duas equipes possuem média de **41 chamados**

## Exercício 2: Qual máquina é mais regular?

```python
maquina_a = [98, 101, 100, 99, 102, 100, 101]
maquina_b = [82, 115, 91, 108, 120, 86, 99]

print(f"{'=' * 30} Dados das máquinas  {'=' * 30}")

print(f"Média da máquina A: {statistics.mean(maquina_a):.2f}")
print(f"Média da máquina B: {statistics.mean(maquina_b):.2f}")

print(f"{'=' * 80}")

print(f"Variância populacional da máquina A: {statistics.pvariance(maquina_a):.2f}")
print(f"Variância populacional da máquina B: {statistics.pvariance(maquina_b):.2f}")

print(f"{'=' * 80}")

print(f"Desvio padrão populacional da máquina A: {statistics.pstdev(maquina_a):.2f}")
print(f"Desvio padrão populacional da máquina B: {statistics.pstdev(maquina_b):.2f}")

print(f"{'=' * 80}")

print(f"Amplitude da máquina A: {calcula_amplitude(maquina_a)}")
print(f"Amplitude da máquina B: {calcula_amplitude(maquina_b)}")

print(f"{'=' * 80}")

print(f"Coeficiente de variação da máquina A: {calcula_coeficiente_variacao(maquina_a):.2f}%")
print(f"Coeficiente de variação da máquina B: {calcula_coeficiente_variacao(maquina_b):.2f}%")
```

### Conclusões:

Ainda que a média seja a mesma para ambas as máquinas, a máquina A é mais regular, pois apresenta menor coeficiente de variação (1.34%) em comparação com a máquina B (14.65%). Isso indica que os valores da máquina A estão mais próximos da média, ou seja, a máquina operou de forma mais consistente, como também é evidenciadp pela amplitude de 4 da máquina A, que é significativamente menor do que a amplitude de 38 da máquina B.

## Exercício 3: Análise dos tempos de entrega?

```python
tempos = [48,32,55,41,37,62,29,45,51,35,70]

print(f"{'=' * 30} Dados dos tempos de entrega  {'=' * 30}")

print(f"Média dos tempos de entrega: {statistics.mean(tempos):.2f}")

print(f"{'=' * 90}")

print(f"Mediana dos tempos de entrega: {statistics.median(tempos)}")

print(f"{'=' * 90}")

print(f"Quartis dos tempos de entrega: ")
quartis = calcular_quartis(tempos)
for i in range(1, 4):
    print(f"{i}º Quartil: {quartis[i - 1]}")

print(f"{'=' * 90}")

print(f"Amplitude dos tempos de entrega: {calcula_amplitude(tempos)}")
print(f"Variância populacional dos tempos de entrega: {statistics.pvariance(tempos):.2f}")
print(f"Desvio padrão populacional dos tempos de entrega: {statistics.pstdev(tempos):.2f}")
print(f"Coeficiente de variação dos tempos de entrega: {calcula_coeficiente_variacao(tempos):.2f}%")

print(f"{'=' * 90}")

print(f"Intervalo interquartil dos tempos de entrega: {quartis[2] - quartis[0]}")
```

### Conclusões:

- Um gestor afirma: **"A maioria das entregas ocorre muito próxima da média."** Os resultados encontrados sustentam essa afirmação?

Os resultados encontrados não sustentam a afirmação do gestor. O coeficiente de variação dos tempos de entrega é de 28.02%, o que indica uma variabilidade significativa em relação à média. Além disso, a amplitude dos tempos de entrega é de 41, mostrando que há uma diferença considerável entre os valores mais baixos e mais altos. O intervalo interquartil de 20 também sugere que os dados estão dispersos, com uma diferença substancial entre o primeiro e o terceiro quartil. Portanto, a maioria das entregas não ocorre muito próxima da média, indicando que há uma variabilidade considerável nos tempos de entrega.
